```python
from tensorflow.keras.models import load_model
```

```python
model = load_model('model.h5')
```

```python
import pickle
```

```python
with open ('ohe.pkl','rb') as file:
    ohe = pickle.load(file)

with open ('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file)

with open ('scalar.pkl','rb') as file:
    scalar = pickle.load(file)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography' : 'France',
    'Gender' : 'Male',
    'Age' : 40,
    'Tenure' : 3,
    'Balance': 60000,
    'NumOfProducts' : 2,
    'HasCrCard' : 1,
    'IsActiveMember' : 1,
    'EstimatedSalary' : 50000
}
```

```python
geo_encoded = ohe.transform([[input_data['Geography']]]).toarray()
```

```python
import pandas as pd
```

```python
geo_encoded_df = pd.DataFrame(geo_encoded,columns=ohe.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df = pd.DataFrame([input_data])
```

```python
input_df['Gender'] = label_encoder_gender.fit_transform(input_df['Gender'])
```

```python
input_df = pd.concat([input_df.drop('Geography',axis=1),geo_encoded_df],axis = 1)
```

```python
input_df
```

```python
input_scaled = scalar.fit_transform(input_df)
```

```python
prediction = model.predict(input_scaled)
```

```python
prediction_prob = prediction[0][0]
```

```python
if prediction_prob>0.5:
    print( "Likely to Churn")
else:
    print( "Not likely to Churn")
```
